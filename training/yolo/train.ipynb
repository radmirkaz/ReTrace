```python
from ultralytics import YOLO

model = YOLO("yolo12n.pt")

results = model.train(
    data="data.yaml",
    imgsz=720, 
    batch=32,
    epochs=50,
    device=0, 
    lr0=3e-4,
    tracker='bytetrack.yaml',
    save_period=10
)
```
